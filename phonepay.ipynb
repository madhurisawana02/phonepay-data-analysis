```
# table 1 aggregated transaction code
import pandas as pd
import json
import os
path = "/content/pulse/data/aggregated/transaction/country/india/state/"
agg_state_list=os.listdir(path)
clm = { 'State':[],'Year':[],'Quater':[],'Transaction_Type':[],'Transaction_count':[],'Transaction_amount':[]}
for i in agg_state_list:
    path_i = path+i+"/"
    agg_yr= os.listdir(path_i)
    for j in agg_yr:
        path_j = path_i+j+"/"
        agg_yr_list=os.listdir(path_j)
        for k in agg_yr_list:
            path_k=path_j+k
            data=open(path_k,'r')
            D=json.load(data)
            for z in D['data']['transactionData']:
                name=z['name']
                count=z['paymentInstruments'][0]['count']
                amount=z['paymentInstruments'][0]['amount']
                clm['Transaction_Type'].append(name)
                clm['Transaction_count'].append(count)
                clm['Transaction_amount'].append(amount)
                clm['State'].append(i)
                clm['Year'].append(j)
                clm['Quater'].append(int(k.strip('.json')))

                #successfullly created a dataFrama
                agg_trans= pd.DataFrame(clm)
```

```
# table 2 aggregated merchant code
import pandas as pd
import json
import os

# Explicitly set path and agg_mer_state_list for merchant data
path ="/content/pulse/data/aggregated/merchant/country/india/state/"
agg_mer_state_list=os.listdir(path)

clm = { 'State':[],'Year':[],'Quarter':[],'Registered_count':[]}
for i in agg_mer_state_list:   #level 1
    path_i = path+i+"/"
    agg_yr= os.listdir(path_i)
    for j in agg_yr:            #level 2
        path_j = path_i+j+"/"
        agg_yr_list=os.listdir(path_j)
        for k in agg_yr_list:     #level 3
            path_k=path_j+k
            data=open(path_k,'r')
            D=json.load(data)
            # Print the keys of D['data']['aggregated'] to inspect the structure
            # print(D['data']['aggregated'].keys()
            Registered_count= D['data']['aggregated']['registeredCount']
            clm['Registered_count'].append(Registered_count)
            clm['State'].append(i)
            clm['Year'].append(j)
            clm['Quarter'].append(int(k.strip('.json')))
#successfully completed dataframe
agg_merchants=pd.DataFrame(clm)
```

```
# table 7 top transaction table code
import pandas as pd
import json
import os
path = "/content/pulse/data/top/transaction/country/india/state/"
top_state_list=os.listdir(path)
clm = { 'State':[],'Districts':[],'Year':[],'Quater':[],'Transaction_count':[],'Transaction_amount':[]}
for i in top_state_list:
    path_i = path+i+"/"
    top_state_yr= os.listdir(path_i)
    for j in top_state_yr:
      path_j = path_i+j+"/"
      top_yr_list=os.listdir(path_j)
      for k in top_yr_list:
        path_k=path_j+k
        data=open(path_k,'r')
        D=json.load(data)
        for z in D['data']['districts']:
          district_name=z['entityName']
          # Directly access type, count, and amount from z['metric']
          transaction_type=z['metric']['type']
          transaction_count=z['metric']['count']
          transaction_amount=z['metric']['amount']

          clm['State'].append(i)
          clm['Districts'].append(district_name)
          clm['Year'].append(j)
          clm['Quater'].append(int(k.strip('.json')))
          clm['Transaction_count'].append(transaction_count)
          clm['Transaction_amount'].append(transaction_amount)

top_transaction=pd.DataFrame(clm)
```
